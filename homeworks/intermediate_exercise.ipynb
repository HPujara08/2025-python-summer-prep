# Q5: 
## Write a function that takes a list and returns a new list that has all the duplicates removed.

Example input and expected output:
- input = `["Michele", "Robin", "Sara", "Michele"]`
- expected output = `['Michele', 'Robin', 'Sara']`


```python
# Solve problem here:

names = ["Michele", "Robin", "Sara", "Michele"]

def removeDup(name_list):
    names = []
    for name in name_list:
        if name not in names:
            names.append(name)
    return names

new_list = removeDup(names)
print(new_list)
```

# Q6: Write a function that takes a list of numbers

# Q8: Write a function that takes a number as input and returns the following:
* If the input is divisible by three, return `'fizz'`
* If the input is divisible by five, return `'buzz'`
* If the input is divisible by three and by five, return `'fizzbuzz'`
* If the input is not divisible by three or five, return `None`.

```python
# Solve Problem fizzbuzz here:
def fizz_buzz(num):
    if num % 3 == 0 and num % 5 == 0:
        return 'fizzbuzz'
    elif num % 3 == 0:
        return 'fizz'
    elif num % 5 == 0:
        return 'buzz'
    else:
        return None
```
